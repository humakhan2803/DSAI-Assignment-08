## Python Project: Online Shopping Cart

**Objective:** Create a simple shopping cart program. A customer can view products, add items to the cart, view the cart and complete checkout.

The program uses dictionaries, loops, conditions, user input and basic calculations.

### 1. Product Details
The products are stored in a nested dictionary. Each product has its name, price and available stock.

```python
products = {
    "P1": {"name": "Shoes", "price": 2500, "stock": 5},
    "P2": {"name": "Shirt", "price": 1500, "stock": 10},
    "P3": {"name": "Bag", "price": 3000, "stock": 4},
    "P4": {"name": "Watch", "price": 4000, "stock": 3},
    "P5": {"name": "Headphones", "price": 2000, "stock": 6}
}

# The cart will store product IDs and their quantities.
cart = {}
```

### 2. View Products
This function displays all products with their prices and available stock.

```python
def view_products():
    print("\n--------------- AVAILABLE PRODUCTS ---------------")
    print("{:<12} {:<15} {:<15} {:<10}".format(
        "Product ID", "Product", "Price (Rs.)", "Stock"
    ))
    print("-" * 55)

    for product_id, details in products.items():
        print("{:<12} {:<15} {:<15} {:<10}".format(
            product_id,
            details["name"],
            details["price"],
            details["stock"]
        ))

    print("-" * 55)
```

### 3. Add Products to Cart
The program checks whether the product ID is correct and whether enough stock is available. If the same product is added again, its quantity is combined with the quantity already in the cart.

```python
def add_to_cart():
    product_id = input("Enter Product ID: ").upper()

    if product_id not in products:
        print("Product not found. Please enter a valid Product ID.")
        return

    try:
        quantity = int(input("Enter quantity: "))

        if quantity <= 0:
            print("Quantity must be greater than zero.")
            return

        current_quantity = cart.get(product_id, 0)
        total_quantity = current_quantity + quantity

        if total_quantity <= products[product_id]["stock"]:
            cart[product_id] = total_quantity
            print(products[product_id]["name"], "added to cart successfully.")
        else:
            available = products[product_id]["stock"] - current_quantity
            print("Not enough stock available.")
            print("You can add", available, "more unit(s) of this product.")

    except ValueError:
        print("Please enter quantity as a whole number.")
```

### 4. View Cart
This function displays the selected products, their quantities, subtotals and the total amount.

```python
def view_cart():
    if len(cart) == 0:
        print("\nYour cart is empty.")
        return

    print("\n---------------- YOUR SHOPPING CART ----------------")
    print("{:<12} {:<15} {:<10} {:<15}".format(
        "Product ID", "Product", "Quantity", "Subtotal (Rs.)"
    ))
    print("-" * 58)

    cart_total = 0

    for product_id, quantity in cart.items():
        name = products[product_id]["name"]
        price = products[product_id]["price"]
        subtotal = price * quantity
        cart_total = cart_total + subtotal

        print("{:<12} {:<15} {:<10} {:<15}".format(
            product_id, name, quantity, subtotal
        ))

    print("-" * 58)
    print("Cart Total: Rs.", cart_total)
```

### 5. Checkout
A 10% discount is given when the cart total is Rs. 5,000 or more. After a successful checkout, the stock is reduced and the cart is cleared.

```python
def checkout():
    if len(cart) == 0:
        print("\nYour cart is empty. Add products before checkout.")
        return

    total = 0

    for product_id, quantity in cart.items():
        total = total + (products[product_id]["price"] * quantity)

    if total >= 5000:
        discount = total * 0.10
    else:
        discount = 0

    final_amount = total - discount

    print("\n================= FINAL BILL =================")
    for product_id, quantity in cart.items():
        name = products[product_id]["name"]
        price = products[product_id]["price"]
        subtotal = price * quantity
        print(name, "x", quantity, "=", "Rs.", subtotal)

    print("----------------------------------------------")
    print("Cart Total:       Rs.", total)
    print("Discount (10%):   Rs.", discount)
    print("Final Amount:     Rs.", final_amount)
    print("==============================================")

    # Reduce stock after successful checkout.
    for product_id, quantity in cart.items():
        products[product_id]["stock"] -= quantity

    cart.clear()
    print("Checkout completed successfully.")
    print("Your cart is now empty. You can start a new order.")
```

### 6. Main Menu
The menu keeps running until the customer selects Exit.

```python
while True:
    print("\n========== ONLINE SHOPPING CART ==========")
    print("1. View Products")
    print("2. Add to Cart")
    print("3. View Cart")
    print("4. Checkout")
    print("5. Exit")

    choice = input("Enter your choice (1-5): ")

    if choice == "1":
        view_products()

    elif choice == "2":
        add_to_cart()

    elif choice == "3":
        view_cart()

    elif choice == "4":
        checkout()

    elif choice == "5":
        print("Thank you for shopping. Goodbye!")
        break

    else:
        print("Invalid choice. Please select a number from 1 to 5.")
```
